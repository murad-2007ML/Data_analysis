```python
import numpy as np
import matplotlib as plt
```

```python
class LinearRegression:

    def __init__(self, learning_rate=0.01, iterations=2000):
        self.learning_rate = learning_rate
        self.iterations = iterations
        self.w = 0
        self.b = 0
        self.cost_history = []

    def predict(self, x):
        x = np.array(x)
        return self.w * x + self.b

    def computing_cost_function(self, x, y):
        m = len(x)
        sum_cost = 0
        for i in range(m):
            prediction = self.predict(x[i])
            sum_cost += (prediction - y[i]) ** 2

        return (1 / (2 * m)) * sum_cost

    def computing_gradient(self, x, y):
        m = len(x)

        d_dw = 0
        d_db = 0
        for i in range(m):
            d_dw += (self.predict(x[i]) - y[i]) * x[i]
            d_db += (self.predict(x[i]) - y[i])

        d_dw /= m
        d_db /= m
        return d_dw, d_db

    def fit(self, x, y):
        
        for i in range(self.iterations):
            d_dw, d_db = self.computing_gradient(x, y)

            self.w -= self.learning_rate * d_dw
            self.b -= self.learning_rate * d_db

            cost = self.computing_cost_function(x, y)
            self.cost_history.append(cost)

            if i % 100 == 0:
                print(f"Iteration {i:4d} | Cost = {cost:.6f}")

    def plot_regression(self, x, y):
        plt.scatter(x, y, label="Data")
        plt.plot(x, [self.predict(i) for i in x], label="Regression Line")
        plt.xlabel("x")
        plt.ylabel("y")
        plt.legend()
        plt.show()

    def plot_cost(self):
        plt.plot(self.cost_history)
        plt.xlabel("Iterations")
        plt.ylabel("Cost")
        plt.title("Cost vs Iterations")
        plt.show()
```

```python
x = np.array([1,2,3,4,5,6,7,8], dtype=float)
y = np.array([3,5,7,9,11,13,15,17], dtype=float)

model = LinearRegression(
    learning_rate=0.01,
    iterations=2000
)

model.fit(x, y)

print("\nFinal Parameters")
print("w =", model.w)
print("b =", model.b)

value = float(input("Enter x: "))
print("Predicted y:", model.predict(value))


model.plot_regression(x, y)
model.plot_cost()
plt.show(block=True)
```
